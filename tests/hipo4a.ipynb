```python
import pandas as pd

data_path = '../results.csv'
df = pd.read_csv(data_path, low_memory=False)
```

```python
# Limpieza: ConvertedCompYearly ✅

# Calculamos el porcentaje de nulos de la columna
print(df['ConvertedCompYearly'].isna().mean() * 100)

# Como ConvertedCompYearly es la variable objetivo, no puedo imputar datos ya que sesgaria el modelo. Si bien hay casi 51.32% de nulos, filtro por aquellos no-nulos aunque pierda mucha información...
df = df.dropna(subset=['ConvertedCompYearly']).copy()
```

```python
# Creación del nuevo dataset ✅

columnas = [
    'ConvertedCompYearly', 
    'Country'
]

df_4a = df[columnas].copy()
```

```python
# Evaluación de la hipótesis ✅

SMVM = 150000
DOLAR = 1500

SMVM_ANUAL_USD = (SMVM * 12) / DOLAR
LIMITE_USD = 3 * SMVM_ANUAL_USD

# Filtramos los datos para desarrolladores que residen en Argentina
df_arg = df_4a[df_4a['Country'] == 'Argentina']

# Calculamos las medidas de tendencia central
mediana_salario_arg = df_arg['ConvertedCompYearly'].median()
media_salario_arg = df_arg['ConvertedCompYearly'].mean()

# Mostramos resultados
print(f"SMVM anual en USD: ${LIMITE_USD}")
print(f"Mediana salario en Argentina (USD): ${mediana_salario_arg}")
print(f"Media salario Argentina (USD): ${media_salario_arg}")

# Conclusión final
if mediana_salario_arg > LIMITE_USD:
    print(f"\nSe confirma la hipótesis. La mediana salarial (${mediana_salario_arg} USD) supera 3 veces el SMVM anual (${LIMITE_USD} USD).")
else:
    print(f"\nSe rechaza la hipótesis. La mediana salarial (${mediana_salario_arg} USD) no alcanza el umbral de 3 veces el SMVM anual (${LIMITE_USD} USD).")
```
